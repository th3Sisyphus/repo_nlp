Menghapus Stopwords

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
stop_words = set(stopwords.words('indonesian'))
example_sent= """Pada mulanya adalah Firman dan Firman itu bersama-sama dengan Allah dan firman itu adalah Allah. Segala sesuatu telah dijadikan oleh Dia dan tanpa Dia tidak ada sesuatupun yang telah jadi dari segala yang telah dijadikan. """
word_tokens = word_tokenize(example_sent.lower())
filtered_sentence = []
for w in word_tokens:
    if w not in stop_words:
        filtered_sentence.append(w)
print(word_tokens)
print(filtered_sentence)
```

Stemming

BoW dan n-gram

```python
import nltk
from nltk import ngrams, FreqDist
kalimat = """Pada mulanya adalah Firman dan Firman itu
            bersama-sama dengan Allah dan firman itu 
            adalah Allah. Segala sesuatu telah dijadikan
            oleh Dia dan tanpa Dia tidak ada sesuatupun 
            yang telah jadi dari segala yang telah dijadikan."""
tokens = nltk.word_tokenize(kalimat.lower())
tokens_b = {}
for b in tokens:
    if b not in tokens_b:
        tokens_b[b] = 1
    else:
        tokens_b[b] += 1
print("token_b \n", tokens_b)
# atau gunakan metode berikut ini:
fdist = {}
fdist = nltk.FreqDist(tokens)
print("this is the content of fdist \n", fdist)
for word, frequency in fdist.most_common(3):
    print(u'{};{}'.format(word, frequency))
```
