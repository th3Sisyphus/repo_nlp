```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
import os
```

# Sampling

```python
path = "D:/Private/Coding/repo_nlp/latihan/dokumen_contoh/fileBunyi"
audio_nonsense = os.path.join(path, 'nonsense.wav')
audio_anger = os.path.join(path, 'anger.wav')
freq_sampling, sinyal_audio = wavfile.read(audio_nonsense)
print("Ukuran sinyal:", sinyal_audio.shape)
print("Tipe data sinyal:", sinyal_audio.dtype)
print("Durasi sinyal:", round(sinyal_audio.shape[0] / float(freq_sampling),2), "seconds")
```

## Visualisasi sampling

```python
sinyal_audio = sinyal_audio / np.power(2,15)
sinyal_audio = sinyal_audio[:100]
time_axis = 1000 * np.arange(0, len(sinyal_audio), 1) / float(freq_sampling)
plt.plot(time_axis, sinyal_audio, color='blue')
plt.xlabel('Waktu (ms)')
plt.ylabel('Amplitudo')
plt.title('Input sinyal audio')
plt.show()
```

# Speech to Text (ASR)

```python
import speech_recognition as sr
import os
```

```python
r = sr.Recognizer()
with sr.AudioFile(audio_nonsense) as source:
    audio = r.record(source)

try:
    print("The audio file contains: " + r.recognize_google(audio))
except sr.UnknownValueError: 
    print("Google Speech Recognition could not understand audio")
except sr.RequestError as e: 
    print("Could not request results from Google Speech Recognition service; {0}".format(e))
```
